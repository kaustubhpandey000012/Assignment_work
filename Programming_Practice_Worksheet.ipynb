```python
# WAP using the two-pointer technique to check whether a given array is palindrome.

def is_palindrome(arr):

    left = 0
    right = len(arr) - 1

    while left < right:
        
        if arr[left] != arr[right]:
            return False
        
        left += 1
        right -= 1

    return True


arr = [1, 2, 3, 2, 1]
print(is_palindrome(arr))


arr = [12121]
print(is_palindrome(arr))


```

```python
# Given a sorted array of integers, WAP using two pointers to find a pair of elements,,whose sum equals
#  a given target value

def find_pair(arr, target):
    left = 0
    right = len(arr) - 1

    while left < right:

        if arr[left] + arr[right] == target:

            return(arr[left], arr[right])

        elif arr[left] + arr[right] < target:
            left = left + 1
        

        else:
            right = right - 1

    return None

arr = [1, 2, 3, 4, 6, 8, 10]
target = 10

print(find_pair(arr, target))
```

```python
#WAP using two-pointer approach to reverse an array in place

def reverse_array(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        arr[left], arr[right] = arr[right], arr[left]

        left += 1
        right -= 1

    return arr

arr = [10, 20, 30, 40, 50]
print(reverse_array(arr))

arr = [1,3,5,7,9,11,13,15]
print(reverse_array(arr))

```
